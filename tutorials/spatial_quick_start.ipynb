# Hisemble: Spatial Modality Quick Start 🗺️

Hisemble is inherently modality-agnostic. In this tutorial, we demonstrate how to apply Hisemble to **Spatial ATAC-seq** data (using the E11.0 mouse brain dataset).

By treating physical spatial coordinates as an independent view alongside the epigenetic features (LSI embeddings), Hisemble effectively fuses spatial contiguity with chromatin accessibility to identify spatially coherent and molecularly distinct tissue domains.

```python
import hisemble
import scanpy as sc
import matplotlib.pyplot as plt

adata = sc.read_h5ad('../demo_data/E11_0.h5ad')
```

```python
hisemble.run_hisemble(
    adata,
    input_keys=['spatial', 'X_lsi'],
    metrics=['gaussian', 'cosine'], 
    k_neighbors=15, 
    resolution=0.45,
    key_added='Hisemble_domains'
)
```

```python
sc.pl.spatial(
    adata, 
    color='Hisemble_domains', 
    title='Hisemble Spatial Domains',
    spot_size=1,
    frameon=False
)
```
